```python
# Exercise 4: Arrange string characters such that lowercase letters should come first

str1 = "PyNaTive"
str_small = ""
str_caps = ""

for i in str1:
  if i.islower():
    str_small += i
  else:
    str_caps += i


print(str_small+str_caps)
```

```python
# Exercise 3: Create a new string made of the first, middle, and last characters of each input string

s1 = "Manohar"
s2 = "manohar"

start = s1[0] + s2[0]
mid = s1[int(len(s1)/2):int(len(s1)/2) + 1] + s2[int(len(s2)/2):int(len(s2)/2) + 1]
end = s1[int(len(s1)) - 1] + s2[int(len(s2)) - 1]

s3 = start + mid + end

print(s3)
```

```python
# Exercise 2: Append new string in the middle of a given string

s1 = "Ault"
s2 = "Kelly"

middle = int(len(s1)/2)
s3 = s1[:middle] +  s2 + s1[middle:]

print(s3)
```

```python
# Exercise 1B: Create a string made of the middle three characters

def get_middle3(a):
  middle = int(len(a)/2)
  mid = a[middle]
  mid = a[middle - 1] + mid + a[middle + 1]
  return mid

print(get_middle3("JhonDipPeta"))
print(get_middle3("JaSonAy"))
print(get_middle3("Required"))

```
